## Dataset 1: IBM HR Analytics Employee Attrition

Es un dataset secundario, publicado por IBM y distribuido en Kaggle. Aunque es ficticio, fue creado por su equipo de ciencia de datos para que se comportara como datos reales de una empresa.
Tiene 1470 registros y 35 columnas: edad, departamento, ingreso mensual, satisfaccion laboral, si hace horas extra, etc. Pesa poco (unos 230 KB) y viene en CSV.

Esta bien documentado, cada columna tiene su descripcion en la pagina de Kaggle y es de los datasets mas usados para practicar clasificacion y EDA en temas de RRHH, asi que hay bastante material de referencia si algo no queda claro.
Sirve para predecir si un empleado se puede ir de la empresa, ver que factores pesan mas en esa decision, o agrupar empleados segun riesgo de renuncia.

```python
import pandas as pd

# Dataset 1: IBM HR Attrition (tabular)
url_hr = "https://raw.githubusercontent.com/pplonski/datasets-for-start/refs/heads/master/employee_attrition/HR-Employee-Attrition-All.csv"
df_hr = pd.read_csv(url_hr)
```
